```python
import random
import matplotlib.pyplot as plt
import torch

from train import classes, folder, load_data, ShapeDataset
from gesture import GestureCNN

_, val, _ = load_data()

path, label = random.choice(val)
image, _ = ShapeDataset([(path, label)])[0]

model = GestureCNN(len(classes))
model.load_state_dict(torch.load(
    folder / "gesture_model.pt",
    map_location="cpu",
    weights_only=True
))
model.eval()

with torch.no_grad():
    prediction = model(image.unsqueeze(0)).argmax(1).item()

plt.imshow(image.squeeze(), cmap="gray")
plt.title(f"Real: {classes[label]} | Predicted: {classes[prediction]}")
plt.axis("off")
plt.show()
```
